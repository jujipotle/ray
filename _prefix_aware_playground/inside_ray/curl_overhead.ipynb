```python
import time
import requests
import subprocess
```

```python
before_curl = time.time()
output = subprocess.check_output(["curl", "-s", "http://localhost:8085/metrics"]).decode("utf-8")
after_curl = time.time()
print(f"Curl time: {after_curl - before_curl}s")
```

```python
session = requests.Session()
before_curl = time.time()
response = session.get("http://localhost:8085/metrics")
output = response.text
after_curl = time.time()
print(f"Curl time: {after_curl - before_curl}s")
```
